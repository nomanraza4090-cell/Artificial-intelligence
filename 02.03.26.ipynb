```python
class Area:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def area(self):
        return self.x * self.y
    
class triangle(Area):
    def __init__(self, x , y):
        super().__init__(x,y)

        def area(self):
            z=super().area()
            ta = z/2
            return ta

k = triangle(1276,3423876)
print(k.area())
        
    
```

```python
# Base class
class Animal:
    def sound(self):
        pass

# Subclass: Dog
class Dog(Animal):
    def sound(self):
        print("Woof! Woof!")

# Subclass: Cat
class Cat(Animal):
    def sound(self):
        print("Meow!")

# Subclass: Cow
class Cow(Animal):
    def sound(self):
        print("Moo!")

# Test the classes
dog = Dog()
cat = Cat()
cow = Cow()

dog.sound()
cat.sound()
cow.sound()
```

```python
# Base class
class Vehicle:
    def move(self):
        pass

# Subclass: Car
class Car(Vehicle):
    def move(self):
        print("Car is driving on the road.")

# Subclass: Bike
class Bike(Vehicle):
    def move(self):
        print("Bike is riding fast!")

# Subclass: Bus
class Bus(Vehicle):
    def move(self):
        print("Bus is moving with passengers.")

# Demonstrate runtime polymorphism using a list
vehicles = [Car(), Bike(), Bus(), Car(), Bike()]

print("=== Runtime Polymorphism Demo ===")
for vehicle in vehicles:
    vehicle.move()
```

```python
import math

# Base class
class Shape:
    def area(self):
        pass

# Subclass: Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):
        return self.length * self.width

# Subclass: Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius ** 2

# Subclass: Triangle
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    
    def area(self):
        return 0.5 * self.base * self.height

# Test the classes
rectangle = Rectangle(5, 3)
circle = Circle(4)
triangle = Triangle(6, 8)

print(f"Rectangle area (5 × 3): {rectangle.area()}")
print(f"Circle area (πr² with r=4): {circle.area():.2f}")
print(f"Triangle area (½ × 6 × 8): {triangle.area()}")
```

```python
# Base class
class Employee:
    def __init__(self, name):
        self.name = name
    
    def get_role(self):
        pass

# Subclass: Manager
class Manager(Employee):
    def get_role(self):
        return f"{self.name} is a Manager - responsible for team leadership and project oversight."

# Subclass: Developer
class Developer(Employee):
    def get_role(self):
        return f"{self.name} is a Developer - responsible for coding and software development."

# Subclass: Designer
class Designer(Employee):
    def get_role(self):
        return f"{self.name} is a Designer - responsible for UI/UX and visual design."


manager = Manager("Alice")
developer = Developer("Bob")
designer = Designer("Carol")

print(manager.get_role())
print(developer.get_role())
print(designer.get_role())
```

```python
# Base class
class Employee:
    def __init__(self, name):
        self.name = name
    
    def calculate_salary(self):
        pass

class FullTimeEmployee(Employee):
    def __init__(self, name, fixed_salary):
        super().__init__(name)
        self.fixed_salary = fixed_salary
    
    def calculate_salary(self):
        return self.fixed_salary

class PartTimeEmployee(Employee):
    def __init__(self, name, hourly_rate, hours):
        super().__init__(name)
        self.hourly_rate = hourly_rate
        self.hours = hours
    
    def calculate_salary(self):
        return self.hourly_rate * self.hours


class Intern(Employee):
    def __init__(self, name, stipend):
        super().__init__(name)
        self.stipend = stipend
    
    def calculate_salary(self):
        return self.stipend


employees = [
    FullTimeEmployee("John", 5000),
    PartTimeEmployee("Sarah", 25, 160),
    Intern("Mike", 1000),
    FullTimeEmployee("Emma", 6000),
    PartTimeEmployee("Alex", 20, 80)
]

print("=== Employee Salary Calculation (Polymorphism) ===")
for employee in employees:
    salary = employee.calculate_salary()
    print(f"{employee.name}: ${salary}")
```

```python
# Base class
class Customer:
    def __init__(self, name, amount):
        self.name = name
        self.amount = amount
    
    def get_discount(self):
        pass

# Subclass: RegularCustomer (5% discount)
class RegularCustomer(Customer):
    def get_discount(self):
        discount = self.amount * 0.05
        return self.amount - discount

# Subclass: PremiumCustomer (10% discount)
class PremiumCustomer(Customer):
    def get_discount(self):
        discount = self.amount * 0.10
        return self.amount - discount

# Subclass: VIPCustomer (20% discount)
class VIPCustomer(Customer):
    def get_discount(self):
        discount = self.amount * 0.20
        return self.amount - discount


customers = [
    RegularCustomer("Alice", 1000),
    PremiumCustomer("Bob", 1000),
    VIPCustomer("Carol", 1000),
    RegularCustomer("David", 5000),
    VIPCustomer("Eva", 5000)
]

print("=== Customer Discount Calculation ===")
for customer in customers:
    final_amount = customer.get_discount()
    discount_percent = {"RegularCustomer": 5, "PremiumCustomer": 10, "VIPCustomer": 20}
    customer_type = customer.__class__.__name__
    discount_pct = discount_percent[customer_type]
    print(f"{customer.name} ({customer_type}, {discount_pct}%): Original: ${customer.amount}, Final Payable: ${final_amount}")
```
