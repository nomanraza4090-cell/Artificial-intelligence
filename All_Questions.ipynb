### # TASK 1: Create variables to store your name, age, and city

```python
name = "John Doe"
age = 25
city = "New York"

print(f"Name: {name} (Type: {type(name).__name__})")
print(f"Age: {age} (Type: {type(age).__name__})")
print(f"City: {city} (Type: {type(city).__name__})")
```

### # TASK 2: Create one variable for each data type

```python
integer_var = 42
float_var = 3.14159
string_var = "Hello, Python!"
boolean_var = True

print(f"Integer: {integer_var} → Type: {type(integer_var)}")
print(f"Float: {float_var} → Type: {type(float_var)}")
print(f"String: {string_var} → Type: {type(string_var)}")
print(f"Boolean: {boolean_var} → Type: {type(boolean_var)}")
```

### # TASK 3: Use type() function to check and display data types

```python
def check_type(variable, var_name):
    print(f"{var_name}: {variable} → Type: {type(variable).__name__}")

check_type(name, "name")
check_type(age, "age")
check_type(integer_var, "integer_var")
check_type(float_var, "float_var")
check_type(string_var, "string_var")
check_type(boolean_var, "boolean_var")
```

### # TASK 4: Program to perform arithmetic operations

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print(f"   {num1} + {num2} = {num1 + num2:.2f} (Addition)")
print(f"   {num1} - {num2} = {num1 - num2:.2f} (Subtraction)")
print(f"   {num1} × {num2} = {num1 * num2:.2f} (Multiplication)")
if num2 != 0:
    print(f"   {num1} ÷ {num2} = {num1 / num2:.2f} (Division)")
else:
    print(f"   {num1} ÷ {num2} = Undefined (Cannot divide by zero)")
```

### # TASK 5: Compare two numbers using comparison operators

```python
a = float(input("Enter first number to compare: "))
b = float(input("Enter second number to compare: "))

print("Comparison Results:")
print(f"   {a} > {b}  : {a > b}")
print(f"   {a} < {b}  : {a < b}")
print(f"   {a} == {b} : {a == b}")
print(f"   {a} != {b} : {a != b}")
print(f"   {a} >= {b} : {a >= b}")
print(f"   {a} <= {b} : {a <= b}")
```

### # TASK 6: Type Conversion

```python
# Integer to Float
int_num = 42
float_converted = float(int_num)
print(f"Integer to Float: {int_num} → {float_converted} ({type(float_converted).__name__})")

# Float to Integer
float_num = 3.14159
int_converted = int(float_num)
print(f"Float to Integer: {float_num} → {int_converted} ({type(int_converted).__name__})")

# String to Integer (if possible)
str_num = "123"
int_from_str = int(str_num)
print(f"String to Integer: '{str_num}' → {int_from_str} ({type(int_from_str).__name__})")

# String to Float
str_float = "45.67"
float_from_str = float(str_float)
print(f"String to Float: '{str_float}' → {float_from_str} ({type(float_from_str).__name__})")
```

### # TASK 7: Display student information

```python
student_name = input("Enter student name: ")
roll_number = int(input("Enter roll number: "))
marks = float(input("Enter marks: "))

print("STUDENT DETAILS:")
print("-"*40)
print(f"Name       : {student_name}")
print(f"Roll Number: {roll_number}")
print(f"Marks      : {marks}")
print(f"Grade      : ", end="")
if marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
elif marks >= 60:
    print("C")
elif marks >= 50:
    print("D")
else:
    print("F")
```

### # TASK 8: Assignment operators

```python
x = 10
print(f"Initial x = {x}")

x += 5
print(f"x += 5  → x = {x}")

x -= 3
print(f"x -= 3  → x = {x}")

x *= 2
print(f"x *= 2  → x = {x}")

x /= 4
print(f"x /= 4  → x = {x}")

x //= 2
print(f"x //= 2 → x = {x}")

x %= 3
print(f"x %= 3  → x = {x}")

x **= 2
print(f"x **= 2 → x = {x}")
```

### # TASK 9: Boolean operators (and, or, not)

```python
bool1 = True
bool2 = False

print(f"Boolean 1: {bool1}")
print(f"Boolean 2: {bool2}")
print(f"bool1 and bool2: {bool1 and bool2}")
print(f"bool1 or bool2 : {bool1 or bool2}")
print(f"not bool1      : {not bool1}")
print(f"not bool2      : {not bool2}")
```

### # TASK 10: Calculate area of a rectangle

```python

length = float(input("Enter length of rectangle: "))
width = float(input("Enter width of rectangle: "))
area = length * width
perimeter = 2 * (length + width)

print(f"Rectangle Calculations:")
print(f"   Length    : {length} units")
print(f"   Width     : {width} units")
print(f"   Area      : {area} square units")
print(f"   Perimeter : {perimeter} units")
```
