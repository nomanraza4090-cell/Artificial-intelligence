```python
class libraryBook:
    def __init__(self , title, author, availablecopies):
        self.title = title
        self.author = author
        self._availablecopies = availablecopies

    def borrowbook(self):
        if self._availablecopies > 0:
            self._availablecopies -= 1
            print(f"You have borrowed '{self.title}' by {self.author}.")
        else:
            print(f"Sorry, '{self.title}' is currently unavailable.")
    def returnbook(self):
        self._availablecopies += 1
        print(f"You have returned '{self.title}' by {self.author}.")
    def showbookdetail(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Available Copies: {self._availablecopies}")

    def show(self):
        print("Welcome to the Library!")
        print(f"Title: {self.title},\nAuthor: {self.author},\nAvailable Books: {self._availablecopies}")
        # print(f"Title: {self.title}")
        # print(f"Author: {self.author}")
```

```python
book1 = libraryBook("To Kill a Mockingbird", "Harper Lee", 3)
book1.show()
print("\n")
book2 = libraryBook("1984", "George Orwell", 2)
book2.show()
print("\n")
book3 = libraryBook("Mr Chips", "John Galsworthy", 5)
book3.show()
book3.borrowbook()
```

```python
class Employee:
    def __init__(self, name, designation, salary):
        self.name = name
        self.designation = designation
        self._salary = salary

    def show_employee_details(self):
        print(f"Name: {self.name}")
        print(f"Position: {self.designation}")
        print(f"Salary: ${self._salary}")

    def increase_salary(self, amount):
        self._salary += amount
        print(f"{self.name}'s salary has been increased by ${amount}. New salary: ${self._salary}")
    
    def showsalary(self):
        print(f"{self.name}'s current salary is: ${self._salary}")

    def show(self):
        print("Employee Details:")
        print(f"Name: {self.name},\nPosition: {self.designation},\nSalary: ${self._salary}")
```

```python
employee1 = Employee("Alice", "Software Engineer", 80000)
employee1.show()
employee1.show_employee_details()
employee1.increase_salary(5000)
employee1.showsalary()
print("\n")
employee2 = Employee("Bob", "Data Scientist", 90000)
employee2.show()
print("\n")
employee2.show_employee_details()
employee2.increase_salary(7000)
employee2.showsalary()
```
