## Break Continue pass

```python
#print numbers untill 5 using  break
num = 1
while num <=10:
    print(num)
    if num ==5:
        break
    num += 1 
```

```python
#skip numeber 3 using continue
num = 1
while num <=5:
    if num ==3:
        num += 1
        continue
    print(num)
    num += 1
```

```python
#loop while pass is placeholder
num = 1
while num <=5:
    pass
    num += 1
```

```python
#stp loop when negative number is eappers
num = 1
while num >=0:
    print(num)
    num = int(input("enter a number"))
    if num < 0:
        print("Negative number entered. Stopping the loop.")
        break
```

```python
#skip even number in loop
num = 1
while num <=10:
    if num % 2 == 0:
        num += 1
        continue
    print(num)
    num += 1
```

```python
#break when sum exceeds 50
total_sum = 0
while True:
    num = int(input("Enter a number to add to the sum: "))
    total_sum += num
    if total_sum > 50:
        print("Sum exceeded 50. Stopping the loop.")
        break
```

```python
#continue for vowels only in string
string = input("Enter a string: ")
vowels = "aeiouAEIOU"
for char in string:
    if char in vowels:
        continue
    print(char)
```

```python
# Infinite loop that exits with break
while True:
    user_input = input("Enter 'exit' to stop the loop: ")
    if user_input.lower() == 'exit':
        print("Exiting the loop.")
        break
    else:
        print("You entered:", user_input)
```

```python
#Skip multiples of 4
num = 1
while num <= 20:
    if num % 4 == 0:
        num += 1
        continue
    print(num)
    num += 1
```

```python
# Use pass inside an if condition
num = int(input("Enter a number: "))
if num > 0:
    pass  
    print("Number is positive")
elif num == 0:
        print("Number is 0")
else:
    print("Number is negative")
```
