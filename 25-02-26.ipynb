```python
#class of resturant in which admin input data and also able to update delete and view date
class Resturant:
    def __init__(self):
        self.menu = {}
    def add_item(self, item_name, price):
        self.menu[item_name] = price
        print(f"{item_name} added to the menu with price Rs {price}.")
    def update_item(self, item_name, new_price):
        if item_name in self.menu:
            self.menu[item_name] = new_price
            print(f"{item_name} price updated to Rs {new_price}.")
        else:
            print(f"{item_name} not found in the menu.")
    def delete_item(self, item_name):
        if item_name in self.menu:
            del self.menu[item_name]
            print(f"{item_name} removed from the menu.")
        else:
            print(f"{item_name} not found in the menu.")
    def view_menu(self):
        if self.menu:
            print("Menu:")
            for item_name, price in self.menu.items():
                print(f"{item_name}: Rs {price}")
        else:
            print("The menu is currently empty.")       
# Example usage
resturant = Resturant()
resturant.add_item("Pizza", 899)
resturant.add_item("Burger", 399)
resturant.add_item("Pasta", 299)
resturant.add_item("Salad", 199)
resturant.add_item("Fries", 149)
resturant.add_item("Soda", 99)
resturant.update_item("Pizza", 1099)
resturant.delete_item("Burger")
resturant.update_item("Ice cream", 300)
resturant.view_menu()
```
