```python
class Rectangle:

    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def __str__(self):
        return f"Rectangle(width={self.width}, height={self.height})"
    
    def area(self):
     
        return self.width * self.height
    
    def __add__(self, other):
        
        if isinstance(other, Rectangle):
            new_width = self.width + other.width
            new_height = self.height + other.height
            return Rectangle(new_width, new_height)
        return NotImplemented
    
    def __eq__(self, other):
      
        if isinstance(other, Rectangle):
            return self.area() == other.area()
        return NotImplemented
    
    def __lt__(self, other):
      
        if isinstance(other, Rectangle):
            return self.area() < other.area()
        return NotImplemented
    
    def __gt__(self, other):
     
        if isinstance(other, Rectangle):
            return self.area() > other.area()
        return NotImplemented

def main():
   
    # Create rectangles
    rect1 = Rectangle(4, 5)
    rect2 = Rectangle(2, 3)
    rect4 = Rectangle(3, 4)  # For comparison example
    
    # Print rectangles and areas
    print("\n--- Rectangle Details ---")
    print(f"Rectangle 1: {rect1}")
    print(f"Area of rect1: {rect1.area()}")
    
    print(f"\nRectangle 2: {rect2}")
    print(f"Area of rect2: {rect2.area()}")
    
    # Operator overloading: +
    print("\n--- Operator Overloading: + ---")
    rect3 = rect1 + rect2
    print(f"rect3 (rect1 + rect2): {rect3}")
    print(f"Area of rect3: {rect3.area()}")
    
    # Operator overloading: ==
    print("\n--- Operator Overloading: == (Area Comparison) ---")
    print(f"Are rect1 and rect2 equal? {rect1 == rect2}")
    
    # Additional comparison
    print(f"Are rect2 and rect4 equal? {rect2 == rect4}")
    print(f"rect4: {rect4} (Area: {rect4.area()})")
    
    # Bonus: Additional operators
    print("\n--- Additional Operators (Bonus) ---")
    print(f"rect1 > rect2? {rect1 > rect2}")
    print(f"rect1 < rect2? {rect1 < rect2}")
    print(f"rect2 == rect4? {rect2 == rect4}")

if __name__ == "__main__":
    main()
```
