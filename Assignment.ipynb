```python
# product discount checker if member and price is 5000 then 20% discount also if not member and price is 5000 then 10% discount othervise no discount
product_price = float(input("Enter product price: "))
is_member = input("Are you a member? (yes/no): ")
if is_member.lower() == "yes" and product_price >= 5000:
    discount = product_price * 0.20
    final_price = product_price - discount
    print(f"Congratulations! You get a 20% discount. Final price: {final_price}")
elif is_member.lower() == "no" and product_price >= 5000:
    discount = product_price * 0.10
    final_price = product_price - discount
    print(f"You get a 10% discount. Final price: {final_price}")
else:
    print("No discount available. Final price:", product_price)
```

```python
name = input("Enter Candidate Name: ")
age = int(input("Enter Age: "))
cgpa = float(input("Enter CGPA: "))
skills = int(input("Enter Skills Count: "))
experience = int(input("Enter Experience (Years): "))
salary = int(input("Enter Expected Salary: "))
communication = int(input("Enter Communication Score (1-10): "))

# ── Reject Conditions ───────────────────────────────────
if age < 18:
    print("Rejected: Candidate must be at least 18 years old.")

elif cgpa < 2.7:
    print("Rejected: CGPA is below minimum requirement.")

elif skills < 3:
    print("Rejected: Minimum 3 skills required.")

else:

    # ── Fraud Detection ──────────────────────────────────
    if salary > 400000 and experience == 0:
        print("⚠ Suspicious Candidate Detected!")
        print("Reason: High salary demand with zero experience.\n")

    # ── Hiring Level Decision ────────────────────────────
    if (cgpa >= 3.3 and skills >= 7 and communication >= 7):
        level = "Software Engineer"

    elif (cgpa >= 3.0 and skills >= 5 and experience >= 1):
        level = "Junior Developer"

    elif (cgpa >= 2.7 and skills >= 3):
        level = "Intern"

    else:
        level = "Not Selected"

    # ── Final Output ─────────────────────────────────────
    print(f"Candidate Name : {name}")
    print(f"Hiring Status  : {level}")
```
