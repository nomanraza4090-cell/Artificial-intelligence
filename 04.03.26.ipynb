```python
import numpy as np
arr = np.array([2,423,6,7,8,3])
print(arr)
arr.ndim
print(arr.ndim)
arr.argmax()
print(arr.argmax)
arr.argmin()
print(arr.argmin)
arr.shape
print(arr.shape)
arr.size
print(arr.size)
arr.dtype
print(arr.dtype)
reshape = arr.reshape(2,3)
print(reshape)
r = np.mean(arr)
print(r)
t = np.median(arr)
print(t)
arr[2:5]
```

```python
a = np.array([2, 3, 4], dtype=float)
print(a)
b = np.array([5, 6, 7])
print(b)
print(b.dtype)
c_unsigned32 = a - b
print('unsigned c:', c_unsigned32, c_unsigned32.dtype)
c_signed32 = a - b.astype(np.int32)
print('signed c:', c_signed32, c_signed32.dtype)
```
