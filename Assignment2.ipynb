# Functions

```python
# Create a function that takes two numbers as parameters and returns their addition.
def add_numbers(num1, num2):
    result = num1 + num2
    return result

print("The sum is:", add_numbers(45, 15))
```

```python
# Write a function that takes a student name and marks as arguments and prints whether the student passed or failed.
def result(name, marks):        
    if marks >= 50:                   
        print(name, "- PASS")         
    else:                             
        print(name, "- FAIL")  

result("Ali", 75)    
result("Sara", 40)
```

```python
# Create a function that accepts a list of numbers and returns the largest number from the list.
def find_largest(numbers):        
    largest = numbers[0]          
    for num in numbers:           
        if num > largest:         
            largest = num         
    return largest                

my_list = [10, 45, 67, 23]                          
print(find_largest(my_list))
```

```python
# Write a function that takes a string as input and counts the number of vowels in it.
def count_vowels(text):               
    vowels = "aeiouAEIOU"             
    count = 0                         
    for letter in text:               
        if letter in vowels:          
            count += 1                
    return count                      

print(count_vowels("Python Programming"))
```

```python
# shopping bill
def calculate_bill(item_name, price, quantity):   
    total = price * quantity                       
    print("Item:", item_name)                      
    print("Total Bill:", total)                    
    return total                                   

calculate_bill("Milk", 120, 3)   
```

#  Lambda Function 

```python
# Write a lambda function to calculate the square of a number.
square = lambda num: num * num 

print(square(5))    
print(square(7)) 
```

```python
# lambda check even or odd
check_even_odd = lambda num: "Even" if num % 2 == 0 else "Odd"

print(4, "is:", check_even_odd(4)) 
print(7, "is:", check_even_odd(7))   
```

```python
# banking system using functions.
account = {
    "name": "Noman Raza",      
    "balance": 45000       
}

def deposit(amount):
    if amount > 0:                             
        account["balance"] += amount          
        print("Deposited:", amount)
        print("New Balance:", account["balance"])
    else:
        print("Invalid amount!")

def withdraw(amount):
    if amount > account["balance"]:            
        print("Insufficient balance!")         
    elif amount <= 0:                          
        print("Invalid amount!")
    else:
        account["balance"] -= amount           
        print("Withdrawn:", amount)
        print("Remaining Balance:", account["balance"])


def check_balance():
    print("Account Holder:", account["name"]) 
    print("Current Balance:", account["balance"]) 

def banking_menu():
    while True:                        
        print("\n--- Banking Menu ---")
        print("1. Deposit")
        print("2. Withdraw")
        print("3. Check Balance")
        print("4. Exit")

        choice = input("Enter choice: ") 

        if choice == "1":
            amt = int(input("Enter deposit amount: "))  
            deposit(amt)                                
        elif choice == "2":
            amt = int(input("Enter withdraw amount: ")) 
            withdraw(amt)                               
        elif choice == "3":
            check_balance()                             
        elif choice == "4":
            print("Thank you! Goodbye.")
            break                                       
        else:
            print("Wrong choice! Try again.")           
print("Welcome to the Banking System!")
banking_menu()
```

```python
# Product Management System
products = {}   

def add_product(name, price):
    products[name] = price              
    print(name, "added successfully!")


def remove_product(name):
    if name in products:                
        del products[name]              
        print(name, "removed!")
    else:
        print(name, "not found!")       


def display_products():
    if len(products) == 0:     
        print("No products available.")
    else:
        print("\n--- Product List ---")
        for name, price in products.items():   
            print(f"{name} : Rs. {price}")     

def total_price():
    total = 0                           
    for price in products.values():     
        total += price                  
    print("Total Price of all products: Rs.", total)


add_product("Rice", 200 , )
add_product("Sugar", 150)
add_product("Oil", 350)
display_products()
total_price()
remove_product("Sugar")
display_products()
```
